Pydantic
Pydantic models provide the richest feature set with field validation, descriptions and nested structures

```python
from pydantic import BaseModel, Field


class Movie(BaseModel):
    title: str = Field(..., description="The title of the movie")
    year: int = Field(..., description="The release year of the movie")
    genre: str = Field(..., description="The genre of the movie")
    director: str = Field(..., description="The director of the movie")
    rating: float = Field(..., description="The rating of the movie (out of 10)")
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model(
    "ollama:qwen3:30b-a3b",
    base_url="http://10.0.0.66:11434",
    reasoning=False,
)
```

```python
model_with_structure = model.with_structured_output(Movie, include_raw=True)
model_with_structure
```

```python
model.invoke("Give me details of a good Christopher Nolan movie")
```

```python
model_with_structure.invoke("Give me details of a good Christopher Nolan movie")
```

```python
response = model_with_structure.invoke("Give me another good Christopher Nolan movie")
print(response)
```

```python
class Actor(BaseModel):
    name: str = Field(..., description="The name of the actor")
    birth_year: int = Field(..., description="The birth year of the actor")


class Member(BaseModel):
    name: str = Field(..., description="The name of the member")
    birth_year: int = Field(..., description="The birth year of the member")


class Movie(BaseModel):
    title: str = Field(..., description="The title of the movie")
    year: int = Field(..., description="The release year of the movie")
    genre: str = Field(..., description="The genre of the movie")
    actors: list[Actor] = Field(..., description="A list of actors in the movie")
    director: Member = Field(..., description="The director of the movie")
    producer: Member = Field(..., description="The producer of the movie")
    crew: list[Member] = Field(
        ..., description="A list of supporting crew members in the movie"
    )
    rating: float = Field(..., description="The rating of the movie (out of 10)")


class MovieList(BaseModel):
    movies: list[Movie] = Field(..., description="A list of movies")


model_with_structure = model.with_structured_output(MovieList)
```

```python
response = model_with_structure.invoke(
    "Give me list of details about the spiderman movies"
)
print(response)
```

TypedDict

```python
from typing import Annotated

from typing_extensions import TypedDict


class MovieDict(TypedDict):
    title: Annotated[str, "The title of the movie"]
    year: Annotated[int, "The release year of the movie"]
    genre: Annotated[str, "The genre of the movie"]
    rating: Annotated[float, "The rating of the movie (out of 10)"]


class MovieListDict(TypedDict):
    movies: Annotated[list[MovieDict], "A list of movies"]


model_with_typed_dict = model.with_structured_output(MovieListDict)
model_with_typed_dict.invoke("Give me details of a good Christopher Nolan movies")
```

Data Classes

```python
from langchain.agents import create_agent

agent = create_agent(
    model,
    response_format=MovieDict,
)

result = agent.invoke(
    {
        "messages": [
            {"role": "system", "content": "You are a helpful assistant."},
            {
                "role": "user",
                "content": "Give me details of a good Christopher Nolan movie",
            },
        ]
    }
)

result["structured_response"]
```

```python
from dataclasses import dataclass


@dataclass
class MovieData:
    title: str
    year: int
    genre: str
    rating: float


agent = create_agent(
    model,
    response_format=MovieData,
)

agent.invoke(
    {
        "messages": [
            {"role": "system", "content": "You are a helpful assistant."},
            {
                "role": "user",
                "content": "Give me details of a good Christopher Nolan movie",
            },
        ]
    }
)
```
