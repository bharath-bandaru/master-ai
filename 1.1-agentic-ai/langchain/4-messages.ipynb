```python
from langchain.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage(content="You are a helpful coding assistant."),
    HumanMessage(content="How do I create a new Python function?"),
]

response = model.invoke(messages)
print(response.content)
```

```python
from langchain.messages import AIMessage

ai_msg = AIMessage(
    content="We have to start with a function definition using the 'def' keyword, followed by the function name and parentheses. Inside the parentheses, we can include any parameters the function will take. After that, we add a colon and indent the next line to write the body of the function. Here's an example:\n\ndef my_function(param1, param2):\n    # Function body\n    result = param1 + param2\n    return result"
)

messages = [
    SystemMessage("You are a helpful coding assistant."),
    HumanMessage("How do I create a new Python function?"),
    ai_msg,
    HumanMessage(
        "Thank you for the explanation! Now write a function that returns Hello World."
    ),
]

response = model.invoke(messages)

for b in response.content_blocks:
    if b["type"] == "reasoning":
        print("🧠", b["reasoning"])

print(response.text)
```

```python
from langchain.messages import ToolMessage

ai_message = AIMessage(
    content=[],
    tool_messages=[
        {
            "name": "get_current_weather",
            "args": {"location": "New York"},
            "reasoning": "I need to get the current weather in New York to answer the user's question.",
            "id": "tool_call_1",
        }
    ],
)

weather_result = "Sunny."
tool_message = ToolMessage(content=weather_result, tool_call_id="tool_call_1")

messages = [
    HumanMessage("What is the current weather in New York?"),
    ai_message,
    tool_message,
]

response = model.invoke(messages)
```

```python
print(response.content)
```
