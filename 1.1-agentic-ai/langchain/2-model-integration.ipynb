```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from langchain.chat_models import init_chat_model

model = init_chat_model("groq:qwen/qwen3.8-27b")
model.invoke("Hello there!")
```

```python
from langchain_groq import ChatGroq

model = ChatGroq(model="qwen/qwen3.8-27b")
model.invoke("Hello there!")
```

Streamings

```python
for chunk in model.stream("Hello there!"):
    print(chunk, end="")
```

Batch

```python
model.batch(["Hello there!", "How are you?"])
```

```python
model.batch(
    ["Hello there!", "How are you?", "how is the weather in New York?"],
    config={"max_concurrency": 2},
)
```
